```python
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import balanced_accuracy_score
from sklearn.model_selection import GridSearchCV
import sklearn
# penalty: Deprecated since version 1.8
sklearn.__version__
```

```python
import joblib
X_train, y_train, X_test, y_test = joblib.load("iris.data")
```

```python
model = LogisticRegression(C=1.0, l1_ratio=0.0, class_weight='balanced', random_state=42, solver='lbfgs')
model.fit(X_train, y_train)

y_train_predict = model.predict(X_train)
y_test_predict = model.predict(X_test)
balanced_accuracy_train = balanced_accuracy_score(y_train, y_train_predict)
balanced_accuracy_test = balanced_accuracy_score(y_test, y_test_predict)
print(f"Ereduaren zehaztasun orekatua - train: {balanced_accuracy_train*100:.2f}%  test: {balanced_accuracy_test*100:.2f}%")
```

```python
grid = {
    'C' : [0.001,0.01,0.1,1,10,100,1000]
}

grid_search = GridSearchCV(
    estimator=LogisticRegression(l1_ratio=0.0, solver='lbfgs', class_weight='balanced', random_state=42),
    param_grid=grid,
    cv = 5,
    scoring='balanced_accuracy'
)

grid_search.fit(X_train, y_train)
print(f'{grid_search.best_params_ = }')
print(f'{grid_search.best_score_*100 = :0.2f}')
```

```python
from sklearn.multiclass import OneVsRestClassifier, OneVsOneClassifier
grid = {
    'estimator__C' : [0.001,0.01,0.1,1,10,100,1000],
    'estimator__l1_ratio' : [0.0, 1.0]
}

model = OneVsRestClassifier(LogisticRegression(solver='liblinear', class_weight='balanced', random_state=42),)

grid_search = GridSearchCV(
    estimator=model,
    param_grid=grid,
    cv = 5,
    scoring='balanced_accuracy'
)

grid_search.fit(X_train, y_train)
print(f'{grid_search.best_params_ = }')
print(f'{grid_search.best_score_*100 = :0.2f}')
```
