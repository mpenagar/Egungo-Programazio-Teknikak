### Konputazio paraleloa eta ausazko zenbakien sorgailuak (RNG)

Programa bat paraleloan exekutatzen denean (hainbat worker erabiliz), normalean worker bakoitzak bere ausazko zenbakiak sortzea nahi da.
* Eman worker guztiei **hazi bera** &rarr; zenbaki-sekuentzia berbera
* Eman worker bakoitzari **hazi desberdina** &rarr; sekuentzien gainjartzea eta korrelazio estatistikoak

Irtenbidea worker bakoitzari RNG **independente** bat ematea da.
1. `parent_RNG = np.random.default_rng(seed)` &rarr; RNG (guraso) objektu bat itzultzen du
1. `children_RNG = parent_RNG.spawn(size)` &rarr; matematikoki independenteak diren $n$ RNG sortzen ditu
1. erabili `children_RNG[i]`

Honako hau lortzen da:
* Independentzia estatistikoa
* Erreproduzigarritasun osoa

#### Arrayen zatitzea (*slice notazioa*)

<br>

NumPy-ren slice notazioak Python-en zerrenden sintaxi bera du

* `x[start:stop:step]`
* Ez badira adierazten, balio lehenetsiak erabiltzen dira:
   * `start=0`
   * `stop=size_of_dimension`
   * `step=1`

````python
x = np.arange(100,110)
print(f'{x[:5]=}')
print(f'{x[5:]=}')
print(f'{x[4:7]=}')
print(f'{x[::2]=}')
print(f'{x[::-1]=}')
print(f'{x[5::-2]=}')
````

````python
x = np.random.randint(1,10,(3,5))
print(f'{x=}')
print(f'{x[1,:]=}')
print(f'{x[1]=}') # baliokidea
print(f'{x[:,1]=}')
print(f'{x[:2,3:]=}')
print(f'{x[0,::-1]=}')
print(f'{x[::-1,1]=}')
````

Python zerrenden sliceek ez bezala, arrayen sliceek **bistak** itzultzen dituzte, ez kopiak:

<br>

````python
x = np.random.rand(3,5)
print(f'{x=}')
y = x[1,::2]
print(f'{y=}')
print('-'*40)
y[0] = 0
print(f'{y=}')
print(f'{x=}')
````

Arrayen edo azpiarrayen (sliceen) kopia esplizituak sor daitezke: `x.copy()`

<br>

````python
x = np.random.rand(3,5)
print(f'{x=}')
y = x[1,::2].copy()
print(f'{y=}')
print('-'*40)
y[0] = 0
print(f'{y=}')
print(f'{x=}')
````

* `vstack` eta `hstack` - 2 (edo gehiago) dimentsioko arrayak
    * `vstack((a1, a2, ...))` == `concatenate((a1, a2, ...), axis=0)`
    * `hstack((a1, a2, ...))` == `concatenate((a1, a2, ...), axis=1)`

<br>

````python
a = np.arange(0,8).reshape((2,4))
b = np.arange(8,16).reshape((2,4))
c = np.hstack((a,b))
d = np.concatenate((a,b), axis=1)
e = np.vstack((a,b))
f = np.concatenate((a,b), axis=0)
print(f'{a=}\n{b=}\n{c=}\n{d=}\n{e=}\n{f=}')
````

* `vstack` eta `hstack` - 1 dimentsioko arrayak
    * `vstack((a1, a2, ...))` == `concatenate((a1[newaxis,:], a2[newaxis,:], ...))`
    * `hstack((a1, a2, ...))` == `concatenate((a1, a2, ...))`

<br>

````python
a = np.arange(0,4)
b = np.arange(4,8)
c = np.hstack((a,b))
d = np.concatenate((a,b))
e = np.vstack((a,b))
f = np.concatenate((a[newaxis,:],b[newaxis,:]))
print(f'{a=}\n{b=}\n{c=}\n{d=}\n{e=}\n{f=}')
````

* `vsplit` eta `hsplit` - 2 (edo gehiago) dimentsioko arrayak
    * `vsplit(a,ii)` == `split(a, ii, axis=0)`
    * `hsplit(a,ii)` == `split(a, ii, axis=1)`
* `hsplit` - 1 dimentsioko arrayak
    * `hsplit(a,ii)` == `split(a, ii)`

<br>

````python
a = np.arange(16).reshape((4,4))
b,c = np.vsplit(a,2)
d,e = np.split(a,2, axis=0)
print(f'{a=}\n{b=}\n{c=}\n{d=}\n{e=}')
````

````python
a = np.arange(16).reshape((4,4))
b,c = np.hsplit(a,2)
d,e = np.split(a,2, axis=1)
print(f'{a=}\n{b=}\n{c=}\n{d=}\n{e=}')
````

````python
a = np.arange(16)
b,c = np.hsplit(a,2)
d,e = np.split(a,2)
print(f'{a=}\n{b=}\n{c=}\n{d=}\n{e=}')
````

## UFunc-ak - Funtzio unibertsalak

Arrayetan elementuz elementuko eragiketak egiteko diseinatutako funtzioak. Ezaugarri nagusiak:

* **Abiadura eta eraginkortasuna**: elementu bakoitzean kalkulu *bera* egiteko arrayak begizta bidez (`for` , `while`) zeharkatu beharrean, optimizatutako funtzioak erabiltzen dira.
* **Broadcasting-a**: forma desberdin baina *bateragarria* duten arrayekin aritzeko aukera.
* **Type casting**: datu mota desberdinekin aritzeko aukera



### Eragiketa aritmetikoak
<br>

````python
x = np.arange(8)
print("     x =", x)
print(" x + 5 =", x + 5)
print(" x - 5 =", x - 5)
print("    -x =", -x)
print(" x * 2 =", x * 2)
print(" x / 2 =", x / 2)
print("x // 2 =", x // 2)
print(" x % 2 =", x % 2)
print("x ** 2 =", x ** 2)
````

Eragile aritmetiko bakoitza, izatez, NumPy-n integratutako funtzio jakin baten estaldura bat (*wrapper*) da

| Eragilea	    | Ufunc baliokidea    | Deskribapena                           |
|---------------|---------------------|---------------------------------------|
|``+``          |``np.add``           |Batuketa (e.g., ``1 + 1 = 2``)         |
|``-``          |``np.subtract``      |Kenketa (e.g., ``3 - 2 = 1``)      |
|``-``          |``np.negative``      |Ezeztapen unarioa (e.g., ``-2``)          |
|``*``          |``np.multiply``      |Biderketa (e.g., ``2 * 3 = 6``)   |
|``/``          |``np.divide``        |Zatiketa (e.g., ``3 / 2 = 1.5``)       |
|``//``         |``np.floor_divide``  |Beheranzko zatiketa osoa (e.g., ``3 // 2 = 1``)  |
|``**``         |``np.power``         |Berreketa (e.g., ``2 ** 3 = 8``)  |
|``%``          |``np.mod``           |Modulua/hondarra (e.g., ``9 % 4 = 1``)|

### Aljebra linealeko eragiketak
Eragile aritmetikoek elementuz elementuko eragiketak adierazten dituzte:

````python
x = np.arange(9).reshape((3,3))
y = np.eye(3)
print(f'{x=}\n{y=}')
print(f'{x+y=}')
print(f'{x*y=}')
````

Aljebra linealeko eragiketak adierazten dituzten eragile bereziak daude:
* `A @ B` edo `np.matmul(A,B)` &rarr; Matrize-biderketa

````python
x = np.arange(9).reshape((3,3))
y = np.eye(3)
print(f'{x=}\n{y=}')
print(f'{x@y=}')
print(f'{np.matmul(x,y)=}')
````

Aljebra linealeko beste eragiketa batzuk:

| "Eragilea"    | Ufunc baliokidea    | Deskribapena                           |
|---------------|---------------------|---------------------------------------|
|``A @ B``      |``np.matmul(A,B)``   |Biderketa         |
|``A.T``        |``np.transpose(A)``  |Iraulketa (bista)   |
|               |``np.linalg.inv(A)`` |Alderantzizkoa          |
|               |``np.linalg.pinv(A)``|Sasi-alderantzizkoa    |
|               |``np.trace(A)``      |Aztarna       |
|               |``np.dot(u, v)``     |Biderkadura eskalarra       |
|               |``np.cross(u, v)``   |Biderkadura bektoriala       |

````python
a = np.array([10,20,30,40,50])
a2 = a.reshape((1,5))
b = np.arange(10).reshape(2,5)
c = np.array([[-1],[-10]])
print(f'{   a = }  {a2 = }')
print(f'{   b = }')
print(f'{   c = }')
print(f'{ a+a = }')
print(f'{ a+3 = }') # broadcasting-a (hedapena)
print(f'{ a+b = }') # broadcasting-a (hedapena)
print(f'{ a+c = }') # broadcasting-a (hedapena)
print(f'{a2+c = }') # broadcasting-a (hedapena)
````

### Broadcasting-aren arauak

$(d_1^x,\dots,d_n^x)$ eta $(d_1^y,\dots,d_m^y)$ dimentsioak dituzten `x` eta `y` arrayak emanda
1. $n \ne m$ bada, dimentsio gutxien dituenaren *shape*-a ezkerretik `1` balioekin betetzen da.
   * $n < m \to (1_1,\dots,1_{m-n},d_1^x,\dots,d_n^x)$
   * $n > m \to (1_1,\dots,1_{n-m},d_1^y,\dots,d_m^y)$
   * Betetzearen ondoren, `x` eta `y`-ren *shape*-ak $(d_1^x,\dots,d_k^x)$ eta $(d_1^y,\dots,d_k^y)$ dira, non $k = \max(n,m)$
1. Bi arrayen formak dimentsioren batean bat ez badatoz, dimentsio horretan 1 tamaina duen arraya hedatu (errepikatu) egiten da bestearen tamainarekin bat etor dadin.
   * Hedapenaren ondoren, $\forall i \;\; d_i^x = 1 \iff d_i^y = 1$
1. Dimentsioren batean tamainak bat ez badatoz ($\exists i \; non \; d_i^x \ne d_i^y$), errore bat sortzen da
   * Errorerik sortzen ez bada, `x` eta `y`-ren *shape*-a $(d_1,\dots,d_k)$ da
3. Dimentsio guztiak bat datoz, emaitzaren *shape*-a $(d_1,\dots,d_k)$ delarik

**Adibidea 1**

````python
M = np.ones((2, 3))
a = np.arange(3)
````

- `M.shape == (2, 3)`
- `a.shape == (3,)`


* 1. araua: betetzea (dimentsio kopurua handitu)
   * `M.shape` &rarr; `(2, 3)`
   * `a.shape` &rarr; `(1, 3)`
* 2. araua: hedatzea (`1` &rarr; `j`)
   * `M.shape` &rarr; `(2, 3)`
   * `a.shape` &rarr; `(2, 3)`
* 3. araua: dimentsioak konprobatu &rarr; **OK**


````python
M + a
````

   * `M.shape` &rarr; `(3, 3)`
   * `a.shape` &rarr; `(3, 3)`
* 3. araua: dimentsioak konprobatu &rarr; **OK**

## Boolearrak eta maskaratzea (*masking*)
* Konparazio-eragileek/ufunc-ek elementuz elementuko boolearrak lortzen dituzte, array boolearrak sortuz
* Array boolearrak (**maskarak**) arrayetan kalkulu selektiboak egiteko erabil daitezke.
* Optimizatutako eragiketa bektorizatuak

````python
x = np.arange(7)
mask = x%3==0
print(f'{        x = }')
print(f'{     mask = }')
print(f'{  x[mask] = }')
print(f'{x[x%3==0] = }')
x[x%3==0] = -10
print(f'{        x = }')
````

`y = x[mask]` beti da **kopia** bat, ez bista bat

````python
x = np.arange(7)
y = x[x%3==0]
print(f'{x = }')
print(f'{y = }')
y[0] = -10
print(f'{x = }')
print(f'{y = }')
````

Konparazio-eragile bakoitza, izatez, NumPy-n integratutako funtzio jakin baten estaldura (*wrapper*) bat da

| Eragilea	    | Ufunc baliokidea    | 
|---------------|---------------------|
|``>``          |``np.greater``       |
|``<``          |``np.less``          |
|``>=``         |``np.greater_equal`` |
|``<=``         |``np.less_equal``    |
|``==``         |``np.equal``         |
|``!=``         |``np.not_equal``     |


Python-en bit mailako eragile logikoak (``&``, ``|``, ``^``, eta ``~``) 
* Elementuz elementuko eragiketa boolearrak (eragile aritmetikoen antzekoak)

<br>

````python
x = np.arange(7)
mask = (x%3==0) & (x<5)
print(f'{                x = }')
print(f'{             mask = }')
print(f'{          x[mask] = }')
print(f'{x[(x%3==0)&(x<5)] = }')
````

Eragile logiko bakoitza, izatez, NumPy-n integratutako funtzio jakin baten estaldura (*wrapper*) bat da:

| Eragilea	    | Ufunc baliokidea   | 
|---------------|--------------------|
|``&``          |``np.bitwise_eta``  |
|  &#124;       |``np.bitwise_or``   |
|``^``          |``np.bitwise_xor``  |
|``~``          |``np.bitwise_not``  |

* `np.any(a)` edo `a.any()` &rarr; `True` | `False` agregatua 
* `np.all(a)` edo `a.all()` &rarr; `True` | `False` agregatua
* `axis` parametroa &rarr; ardatz jakin baten gaineko agregazioa

<br>

````python
x = np.arange(10).reshape((2,5))
print(f'{x=}')
print(f'{x>7=}')
print(f'{np.any(x>7)=}')
print(f'{(x>7).any()=}')
print(f'{np.all(x>7)=}')
print(f'{(x>7).all()=}')
print(f'{np.any(x>7, axis=0)=}')
print(f'{np.any(x>7, axis=1)=}')
````

Array boolearrak batu daitezke
* `True==1` , `False==0`
* `np.sum(a)` edo `a.sum()` &rarr; `True` | `False` agregatua
* `axis` parametroa &rarr; ardatz jakin baten gaineko agregazioa
* **EZ ERABILI** Pythoneko `sum` built-in funtzioa

<br>

````python
x = np.arange(10).reshape((2,5))
print(f'{x=}')
print(f'{x>7=}')
print(f'{np.sum(x>7)=}')
print(f'{(x>7).sum()=}')
print(f'{np.sum(x>7, axis=0)=}')
print(f'{np.sum(x>7, axis=1)=}')
````

## *Fancy Indexing*
* Arrayetako elementu anitzerara iristeko, indize arrayak erabil daitezke
* Datuen **kopia** bat itzultzen du (slicing-aren *view*-aren ordez)

<br>

````python
a = np.random.randint(100,size=(10,))
ii1 = [1, 8, 3, 5]
ii2 = np.array([1, 8, 3, 5])
print(f'{     a = }')
print(f'{a[ii1] = }')
print(f'{a[ii2] = }')
````

*Fancy Indexing*-a dimentsio anitzeko arrayekin erabil daiteke
* Arrayetako elementu anitzerara iristeko, indize arrayak erabil daitezke
* **Hautaketa sekuentziala**: dimentsio bakarreko array bat itzultzen du

<br>

````python
a = np.arange(15).reshape((3,5))
ii = [1, 0, 2, 2]
jj = [1, 4, 3, 2]
print(f'{    a = }')
print(f'{a[ii,jj] = }')
````

*Fancy Indexing*-a indize soilekin konbina daiteke
* *Indizeen broadcasting* moduko bat

<br>

````python
a = np.arange(16).reshape((4,4))
ii = [1, 3, 2]
print(f'{    a = }')
print(f'{a[1,ii] = }')
print(f'{a[ii,2] = }')
````

Azpiarrayak lor daitezke
* *Fancy Indexing*-a lehen dimentsioan soilik erabiltzean &rarr; errenkadak
* *slicing*-arekin konbinatuz &rarr; zutabeak

<br>

````python
a = np.arange(15).reshape((3,5))
ii = [2, 0]
jj = [1, 4, 3]
print(f'{    a = }')
print(f'{a[ii] = }')
print(f'{a[ii,:] = }')
print(f'{a[:,jj] = }')
````
