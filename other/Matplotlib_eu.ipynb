````python
import matplotlib as mpl
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np
import pandas as pd

# Ezarri irudiaren tamaina lehenetsia hazbetetan (zabalera, altuera)
mpl.rcParams['figure.figsize'] = (5.33,4)
# Ezarri ardatzen eta marka-etiketen letra-tamaina lehenetsia puntutan
mpl.rcParams['axes.labelsize'] = 10  # Adibidea: 14 puntu
mpl.rcParams['xtick.labelsize'] = 8  # Adibidea: 12 puntu x ardatzeko marketarako
mpl.rcParams['ytick.labelsize'] = 8  # Adibidea: 12 puntu y ardatzeko marketarako
````

## Estiloak

* Aurrez definitutako atributu bisualen bilduma
* `plt.style.library` &rarr; eskuragarri dauden estiloak eta haien parametroak
   * `plt.style.available` == `sorted(plt.style.library.keys())` 
* `plt.style.use('stylename')` &rarr; estilo globala ezarri
   * Testuinguru-kudeatzailea (`with` *context manager*) erabili estiloa lokalki aldatzeko:
   * ```python
     with plt.style.context('stylename'):
         make_a_plot()
     ```

````python
def hist_and_lines(title="Estilo lehenetsia"):
    np.random.seed(0)
    fig, ax = plt.subplots(1, 2, figsize=(6, 2.5))
    ax[0].hist(np.random.randn(1000))
    for i in range(3):
        ax[1].plot(np.random.rand(10))
    ax[1].legend(['a', 'b', 'c'], loc='lower left')
    fig.suptitle(title)
````

````python
hist_and_lines()
````

````python
plt.style.available
````

````python
for style in plt.style.available[:7] :
    with plt.style.context(style):
        hist_and_lines(f'{style=}')
hist_and_lines()
````

### Seaborn estiloak

* Bost estilo: `darkgrid`, `whitegrid`, `dark`, `white`, and `ticks`
* `sns.set_style(stylename)` &rarr; ezarri estilo globala
   * Erabili testuinguru-kudeatzailea (`with` *context manager*) estiloa lokalki aldatzeko:
   * ```python
     with sns.axes_style(stylename):
         make_a_plot()
     ```

````python
for style in ['darkgrid', 'whitegrid', 'dark', 'white', 'ticks'] :
    with sns.axes_style(style):
        hist_and_lines(f'{style=}')
hist_and_lines()
````

* `sns.set()` == `sns.set_theme()` &rarr; sns.set_style('darkgrid') eta gehiago

````python
sns.set()
mpl.rcParams['figure.figsize'] = (5.33,4)
mpl.rcParams['axes.labelsize'] = 10  # Adibidea: 14 puntu
mpl.rcParams['xtick.labelsize'] = 8  # Adibidea: 12 puntu x ardatzeko marketarako
mpl.rcParams['ytick.labelsize'] = 8  # Adibidea: 12 puntu y ardatzeko marketarako
hist_and_lines("Estilo lehenetsi berria")
````

## Histogramak
* `plt.hist(x, bins=, density=, histtype=)` &rarr; histograma bat

````python
x = np.random.randn(10000)
plt.hist(x);
````

````python
plt.hist(x, bins=40, density=True, histtype='stepfilled');
````

````python
density, bins, patches = plt.hist(x, bins=40, density=True, histtype='stepfilled')
# Kalkulatu tarteen erdigunea
bin_centers = (bins[:-1] + bins[1:]) / 2
# Marraztu probabilitate-dentsitatea lerro gisa
plt.plot(bin_centers, density);
````

````python
x1 = np.random.normal(0, 0.8, 10000)
x2 = np.random.normal(-2, 1, 10000)
x3 = np.random.normal(3, 2, 10000)
kwargs = dict(bins=40, density=True, histtype='stepfilled', alpha=0.4)
plt.hist(x1, **kwargs)
plt.hist(x2, **kwargs)
plt.hist(x3, **kwargs);
````

## Seaborn
* Matplotlib Pandas baino hamarkada bat lehenago sortu zen
   * Ez zen Pandas-ekin erabiltzeko diseinatu (eguneratzen ari da)
* Seaborn-ek Matplotlib-en gaineko API bat eskaintzen du eta Pandas-ekin integratzen da
  * Tresna estatistikoak: erregresioak, erroreak eta banaketak
  * Aldagai anitzeko bistaratze sinplea

````python
a = np.random.multivariate_normal(mean=[0, 1], cov=[[5, 2], [2, 2]], size=2000)
df = pd.DataFrame(a, columns=['x', 'y'])
df.head()
````

Bi histograma marraztea (<u>tarte-zabalerak</u> desberdinak dira) 

````python
plt.hist(df['x'], bins=40, alpha=0.4, label='x')
plt.hist(df['y'], bins=40, alpha=0.4, label='y')
plt.legend();
````

Seaborn-ek DataFrame-ko aldagaien histogramak marraz ditzake (<u>tarte-zabalera berdinekin</u>):

````python
sns.histplot(df, bins=40);
````

DataFrame-ko aldagaien kernel-dentsitatearen estimazioa:

````python
sns.kdeplot(df, fill=True);
````

Kernel-dentsitatearen estimazioaren bi dimentsioko bistaratzea:

````python
sns.kdeplot(df, x='x', y='y');
````

Banaketa bateratua eta marjinalak batera:

````python
with sns.axes_style('white'):
    sns.jointplot(df, x='x', y='y', kind='kde', height=4);
````

Gauza bera, baina hexagonoetan oinarritutako histograma erabiliz:

````python
with sns.axes_style('white'):
    sns.jointplot(df, x='x', y='y', kind='hex', height=4);
````

### Aldagai anitzeko bistaratzea

**Iris** datu multzoa: hiru iris-espezieren petaloen eta sepaloen neurketak: 

````python
iris = sns.load_dataset("iris")
print(iris.shape)
iris.head()
````

Datuen korrelazio-matrizea:

````python
iris.drop(['species'], axis=1).corr()
````

Korrelazio-matrizearen bero-mapa:

````python
c = iris.drop(['species'], axis=1).corr()
sns.heatmap(c, cmap="Blues")
````

`setosa` espeziearen korrelazio-matrizearen bero-mapa:

````python
c = iris[iris['species']=='setosa'].drop('species', axis=1).corr()
sns.heatmap(c, cmap="Blues")
````

Gauza bera, baina `species` aldagaia indize gisa erabiliz:

````python
c = iris.set_index('species').loc['setosa'].corr()
sns.heatmap(c, cmap="Blues")
````

Espezie bakoitzaren korrelazio-matrizea:

````python
iris.groupby('species').corr()
````

Espezie bakoitzaren korrelazio-matrizearen bero-mapa:

````python
sns.heatmap(iris.groupby('species').corr(),cmap="Blues")
````

Aldagai bikoteen sakabanatze-diagramak eta KDE dituen grafikoa: 

````python
sns.pairplot(iris, hue='species', height=1.5);
````
