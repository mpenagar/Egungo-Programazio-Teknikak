# Matplotlib

Ariketa multzo honek **matplotlib** liburutegiaren oinarrizko eragiketak lantzen ditu, Ikaskuntza Automatikoan (ML) erabilgarriak diren datuen esplorazioa, ereduen ebaluazioa eta abarren gaitasunak landuz.

```python
import numpy as np
import matplotlib as mpl
import matplotlib.pyplot as plt
import seaborn as sns

# Ezarri irudiaren tamaina lehenetsia hazbetetan (zabalera, altuera)
mpl.rcParams['figure.figsize'] = (5.33,4)
# Ezarri ardatzen eta marka-etiketen letra-tamaina lehenetsia puntutan
mpl.rcParams['axes.labelsize'] = 10  # Adibidea: 10 puntu
mpl.rcParams['xtick.labelsize'] = 8  # Adibidea: 8 puntu x ardatzeko marketarako
mpl.rcParams['ytick.labelsize'] = 8  # Adibidea: 8 puntu y ardatzeko marketarako
```

##  Errendimendu eta Galera Kurbak

**1 - Galera-kurba (Loss Curve):** Demagun eredu batek duen galera (*loss*) funtziaren kurbak $y_{train} = e^{-0.2x}$ itsura duela, $x$ ardatzeko balioak entrenamendu garaiak direlarik (*epochs*, iterazioak direla suposatu, $[1,50]$ tartean). Marraztu ezazu kurba hori ([plt.plot](https://matplotlib.org/stable/api/_as_gen/matplotlib.pyplot.plot.html)), ezarri titulu bat eta izendatu ardatzak (*epochs* eta *loss*). Adibidez:



```python
x = np.arange(1,51)
y = np.exp(-0.2*x)
plt.axes().set(xlabel='epoch', ylabel='loss', title='Loss Curve')
plt.plot(x, y);
```

**2 - Entrenamendua vs. Balioztatzea (Overfitting):** Aurreko adibidearekin jarraituz, demagun balioztatze-galera (*validation loss*, test-galera dela suposatu) kurbak $y_{val} = e^{-0.15x} + 0.01x$ itsura duela. Marraztu ezazu grafiko berean entrenamenduko galera eta balioztatze-galera, ezarri titulu bat,  ardatzak izendatu eta gehitu legenda bi lerroak bereizteko. Adibidez:



```python
x = np.arange(1,51)
y_train = np.exp(-0.2*x)
y_valid = np.exp(-0.15*x) + 0.01 * x
plt.axes().set(xlabel='epoch', ylabel='loss', title='Loss Curve')
plt.plot(x, y_train, label='train')
plt.plot(x, y_valid, label='valid')
plt.legend();
```

**3 - Lerro-estiloak metrika anitzetan:** Demagun eredu bat 50 *epoch*-etan entrenatzean bi metrika erabili direla, Zehaztasuna (*Accuracy*) eta Prezisioa (*Precision*) eta bi metrika horien kurbek $y_{acc}=0.8-e^{-0.2x}$ eta $y_{prec}=\max \{0 ,\, 0.7-e^{-0.1x}\}$ itsura dutela. Irudikatu *epoch*-etan zehar lortutako *Accuracy* eta *Precision* metrikak. Erabili lerro jarraitua bata marrazteko eta puntu bidezkoa (`--`) bestearentzat, kolore desberdinekin. Adibidez:



```python
x = np.arange(1,51)
y_acc = 0.8-np.exp(-0.2*x)
y_prec = np.max([np.zeros(len(x)),0.7-np.exp(-0.1*x)], axis=0)
plt.axes().set(xlabel='epoch', ylabel='Acc / Prec', title='Performance', ylim=(0,1))
plt.plot(x, y_acc, '-', label='Accuracy')
plt.plot(x, y_prec, '--', label='Precision')
plt.legend();
```

## Sakabanatze-Diagramak eta Ezaugarrien Espazioa (Scatter Plots)


**4 - Bi ezaugarritako sakabanatze-diagrama:** Sortu ezazu $n=1000$ laginez eta bi ezaugarriz osotutako $(n \times 2)$ tamainako $X$ ezaugarri matrize bat. Laginak auzazki sortu $\mathcal{N}\left( \boldsymbol{\mu}, \boldsymbol{\Sigma} \right)$ banaketaren arabera, non

$$\boldsymbol{\mu} = \begin{pmatrix} 0 \\ 15 \end{pmatrix}, \quad \boldsymbol{\Sigma} = \begin{pmatrix} \sigma_x^2 & 0 \\ 0 & \sigma_y^2 \end{pmatrix} = \begin{pmatrix} 1 & 0 \\ 0 & 10 \end{pmatrix}$$

Ondoren $X$ ezaugarri matrizeko datu-puntuen sakabanatze-diagrama bat ([`plt.scatter`](https://matplotlib.org/stable/api/_as_gen/matplotlib.pyplot.scatter.html)) marraztu. Adibidez:



```python
n = 1000
X = np.zeros((n,2))
X[:,0] = np.random.normal(loc=0, scale=1, size=X.shape[0])
X[:,1] = np.random.normal(loc=15, scale=10, size=X.shape[0])
plt.axes().set(xlabel='$x_1$', ylabel='$x_2$', title='Scatter Plot')
plt.scatter(X[:,0], X[:,1]);
```

**5 - Klaseak koloreztatzea (`c` eta `cmap`):** Aurreko ezaugarri matrizea erabiliz, sortu ezazu $n$ luzerako $\boldsymbol{y}$ klase/etiketa bektore bat, zeinaren balioak $X$-ko lehenengo ezaugarriaren zeinua adierazten duen, $1$ positiboa denean eta $0$ negatiboa denean. $X$ ezaugarri matrizeko datu-puntuen sakabanatze-diagrama bat ([`plt.scatter`](https://matplotlib.org/stable/api/_as_gen/matplotlib.pyplot.scatter.html)) marraztu eta erabili ezazu $\boldsymbol{y}$ bektorea puntuen kolore adierazle gisa `c=y` argumentuaren bidez. Froga itzazu [*kolore paleta*](https://matplotlib.org/stable/users/explain/colors/colormaps.html) ezberdinak (`cmap='viridis'`, `cmap='coolwarm'`, `cmap='rainbow'`, `cmap='cool'`...). Adibidez:



```python
y = 0 + (X[:,0] >= 0)
plt.axes().set(xlabel='$x_1$', ylabel='$x_2$', title='Scatter Plot')
plt.scatter(X[:,0], X[:,1], c=y, cmap='viridis');
```

**6 - Hirugarren ezaugarria dimentsionatzea:** Aurreko $\mathcal{N}\left( \boldsymbol{\mu}, \boldsymbol{\Sigma} \right)$ banaketa berdinarekin sortu ezazu orain $n=50$ tamainako $X$ ezaugarri matrize bat. Gehitu iezaiozu hirugarren ezaugarri bat, $[0,1000)$ arteko auzazko balioz osotua. Marraztu lehenengo bi ezaugarrien sakabanatze diagrama eta hirugarren ezaugarria `s` parametroari esleitu, puntu bakoitzaren magnitudea bistaratzeko. Jokatu ezazu `alpha` parametroarekin, zeinak puntuen gardentasuna kontrolatzen duen (`alpha=0` guztiz gardena eta `alpha=1.0` guztiz opakua). Adibidez:



```python
n = 50
X = np.zeros((n,3))
X[:,0] = np.random.normal(loc=0, scale=1, size=X.shape[0])
X[:,1] = np.random.normal(loc=15, scale=10, size=X.shape[0])
X[:,2] = 1000*np.random.rand(X.shape[0])

plt.axes().set(xlabel='$x_1$', ylabel='$x_2$', title='Scatter Plot with Sizes')
plt.scatter(X[:,0], X[:,1], s=X[:,2], alpha=0.4);
```

**7 - Erregresio-lerroa datuen gainean:**

Demagun erregresio eredu lineal sinple bat dugula: $y = 0.7x + 3.5$ . Sortu ezazu $[0,10]$ tartean linealki banatua dagoen $n=300$ tamainako $\boldsymbol{x}$ bektore bat. $\boldsymbol{x}$ bektoreko balioekin, sortu ezazu $\boldsymbol{y}$ bektore bat, zeinetan ereduak aurresandako $y$ balioa eta $[-1,1]$ tarteko auzazko balio baten batura egongo den ($\boldsymbol{x}$-ko puntu bakoitzak, ausazko balio ezberdin bat izango du). Marraztu $\boldsymbol{x},\boldsymbol{y}$ datu-puntu sakabanatuak eta, horien gainean, ereduaren erregresio-lerro zuzena, $y = 0.7x + 3.5$. Jokatu ezazu `scatter` funtzioaren `marker='<ZEINU_BAT>'` eta `c='<KOLORE_IZEN_BAT>'` parametroekin eta `plot` funtzioaren `linewidth=<ILARA_ZABALERA>` parametroarekin irudi atseginago bat lortzeko. Adibidez:



```python
x = np.linspace(0, 10, 300)
y = 0.7 * x + 3.5
y2 = y + 2*np.random.rand(x.size) - 1

plt.axes().set(xlabel='x', ylabel='y', title='Linear Regression')
plt.scatter(x, y2, marker='.', c='grey')
plt.plot(x, y, linewidth=5);
```

## Datu-Banaketak eta Ezaugarrien Azterketa (EDA)

**8 - Ezaugarri baten histograma:** Sortu ezazu ezaugarri jarraitu bakarra duen $n=1000$ tamainako $\boldsymbol{x}$ bektore bat ($\mathcal{N}(\mu,\sigma)$ banaketa bat erabiliz) eta marraztu bere banaketa ([`plt.hist`](https://matplotlib.org/stable/api/_as_gen/matplotlib.pyplot.hist.html)) 30 `bins` erabiliz. Adibidez:



```python
n = 1000
x = np.random.normal(loc=3, scale=2, size=n)
plt.axes().set(xlabel='x', ylabel='count', title='Histogram')
plt.hist(x, bins=30);
```

**9 - Klaseen araberako banaketa gainjarria:** Sortu aurreko ariketako moduko bi $\boldsymbol{x_{0}}$ eta $\boldsymbol{x_{1}}$ bektore (bakoitza batazbesteko eta desbiderapen estandar ezberdinarekin). Sortu itzazu $\boldsymbol{y_{0}}=[0,0,\dots,0]$ eta $\boldsymbol{y_{1}}=[1,1,\dots,1]$ bi etiketa bektore. Nahastu datu guztiak, $\boldsymbol{x}$ eta $\boldsymbol{y}$ bakarra sortuz, baina $x_i$ bakoitzari dagokion $y_i$ etiketa zuzena izaten jarrai dezan. $\boldsymbol{x}$ eta $\boldsymbol{y}$ bektoreetatik habiatuz, marraztu bi klase ezberdinen ezaugarrien histogramak grafiko berean, gardentasuna erabiliz (`alpha=0.5`) bi banaketak konparatu ahal izateko. Adibidez:



```python
n = 1000
x0 = np.random.normal(loc=3, scale=2, size=n)
x1 = np.random.normal(loc=-1, scale=1, size=n)
y0 = np.zeros(n)
y1 = np.ones(n)

x = np.concatenate((x0, x1))
y = np.concatenate((y0, y1))

index = np.arange(2*n)
np.random.shuffle(index)
x = x[index]
y = y[index]

mask_0 = y == 0
mask_1 = y == 1
x0,y0 = x[mask_0],y[mask_0]
x1,y1 = x[mask_1],y[mask_1]

plt.axes().set(xlabel='x', ylabel='count', title='2 Class Histograms')
plt.hist(x0, bins=30, alpha=0.5, label='$class_0$')
plt.hist(x1, bins=30, alpha=0.5, label='$class_1$')
plt.legend();
```

**10 - Klaseen desoreka (Bar Plot):** Sortu ezazu $n=1000$ tamainako auzazko $\boldsymbol{y}$ etiketa bektorea, $\{0,1\}$ balioez osatua, etiketa bakoitza $1$ izateko probabilitatea $0.3$ delarik. Sortu barra-diagrama bat ([`plt.bar`](https://matplotlib.org/stable/api/_as_gen/matplotlib.pyplot.bar.html)) klase bakoitzaren instantzia kopurua erakusteko (*Class Imbalance* hautemateko). Adibidez:



```python
n = 1000
#y = 0 + (np.random.rand(n) > 0.7)
y = np.random.choice([0, 1], size=n, p=[0.7, 0.3])
height = [n-np.sum(y),np.sum(y)]
plt.axes().set(xlabel='class', ylabel='count', title='Class Balance')
plt.bar(["0","1"], height);
```

**11 - Outlier-ak hautematea (Boxplot):** Sortu ezazu $n=1000$ adibide eta bi ezaugarri dituen $(n \times 2)$ tamainako $X$ datu matrizea, ezaugarrientzat $\mathcal{N}(\mu_1,\sigma_1)$ eta $\mathcal{N}(\mu_1,\sigma_2)$ banaketak erabiliz. Auzazki $k=10$ adibide aukeratu eta lehenengo ezaugarriko balioak aldatu, $\mathcal{N}(\mu_1,5\sigma_1)$ banaketa erabiliz. Auzazki $k=10$ adibide aukeratu eta bigarren ezaugarriko balioak aldatu, kasu honetan $\mathcal{N}(\mu_2,3\sigma_2)$ banaketa erabiliz.

Marraztu ezazu *whisker* tramak dituen kutxa bat ([`plt.boxplot`](https://matplotlib.org/stable/api/_as_gen/matplotlib.pyplot.boxplot.html)), bi ezaugarrien banaketa eta balio atipikoak (*outliers*) ikusteko.

```python
n = 1000
k = 10
mean_1 = 0
std_1 = 1
mean_2 = 5
std_2 = 4

X = np.empty((n,2))
X[:,0] = np.random.normal(loc=mean_1, scale=std_1, size=n)
X[:,1] = np.random.normal(loc=mean_2, scale=std_2, size=n)

index = np.random.choice(np.arange(n), k, replace=False)
X[index,0] = np.random.normal(loc=mean_1, scale=5*std_1, size=k)
index = np.random.choice(np.arange(n), k, replace=False)
X[index,1] = np.random.normal(loc=mean_2, scale=3*std_2, size=k)
plt.boxplot(X);
```

## Azpigrafikoak eta Sare-Egiturak (Subplots)


**12 - Ezaugarri anitzen panela ($2 \times 2$ grid):** Sortu ezazu $n=1000$ adibide eta lau ezaugarri dituen $(n \times 4)$ tamainako $X$ datu matrizea, ezaugarri bakoitzarentzat $\mathcal{N}(\mu_i,\sigma_i) \,,\, i \in [0,3]$ banaketa erabiliz.
Sortu $2 \times 2$ saretako azpigrafiko panel bat ([`plt.subplots`](https://matplotlib.org/stable/api/_as_gen/matplotlib.pyplot.subplots.html)), non azpigrafiko bakoitzean ezaugarri baten histograma erakusten den.

```python
n = 1000
X = np.empty((n,4))
means = np.empty(4)
stds = np.empty(4)

for i in range(4) :
    means[i] = 10 * np.random.random()
    stds[i] = np.random.random()
    X[:,i] = np.random.normal(loc=means[i], scale=stds[i], size=n)

fig, axs = plt.subplots(2, 2, figsize=(2*5.33, 2*4))
axs = axs.reshape(4)
for i in range(4) :
    axs[i].set(xlabel='x', ylabel='count', title=f'mean={means[i]:.2f} std={stds[i]:.2f}')
    axs[i].hist(X[:,i])
plt.tight_layout()
```

**13 - Ardatz bikoitzeko grafikoa (`twinx`):** 

Demagun 50 *epoch*-etan entrenatu den eredu baten galera funtziaren (*loss*) eta zehaztasunaren (*accuracy*) kurbek $y_{loss} = e^{-0.2x}$ eta $y_{acc}=0.8-e^{-0.1x}$ itsura dutela. Irudikatu ezazu galera (ezkerreko Y ardatza) eta zehaztasuna (eskuineko Y ardatza) grafiko bakar batean [`ax.twinx()`](https://matplotlib.org/stable/api/_as_gen/matplotlib.pyplot.twinx.html) erabiliz.


```python
x = np.arange(1,51)
y_loss = np.exp(-0.2*x)
y_acc = 0.9-np.exp(-0.1*x)

fig, ax1 = plt.subplots()
ax1.set(xlabel='epoch', ylabel='loss', title='Training')
ax1.plot(x, y_loss, label='loss')

ax2 = ax1.twinx()
ax2.set(ylabel='accuracy', ylim=(0,1))
ax2.plot(x, y_acc, 'g', label='accuracy')

handles1, labels1 = ax1.get_legend_handles_labels()
handles2, labels2 = ax2.get_legend_handles_labels()
ax1.legend(handles1 + handles2, labels1 + labels2, loc='center right')
```

## Matrizeak, Bero-Mapak eta Irudiak

**14 - Nahasmen Matrizea (Confusion Matrix):** [*Nahaste-matrize*](https://eu.wikipedia.org/wiki/Nahaste-matrize) batek sailkatzaile batek egindako sailkatze zuzenak eta okerrak adierazten ditu. Sortu $3 \times 3$ tamainako Nahaste-matrize bat eta irudikatu ezazu [`plt.imshow`](https://matplotlib.org/stable/api/_as_gen/matplotlib.pyplot.imshow.html) erabiliz (colore-mapa gris alderantzizkoa erabili, `cmap='gray_r'`, balio handiek kolore ilun bat faso dezaten). Gelaxka bakoitzaren barruan, testu bidez, adierazi balio numeriko bakoitza, `axis` objetuen [`text`](https://matplotlib.org/stable/api/_as_gen/matplotlib.axes.Axes.text.html) funtzioa erabiliz:

`ax.text(0, 2, 'kaixo', ha='center', va='center')` -ek adibidez 3. ilarako 1. zutabeko gelaxkan (x=0,y=2 argumentuak matrizeko j=2,i=0 indizeak lirateke) `'kaixo'` idatziko du (horizontalki eta bertikalki zentratua). 

```python
M = np.array([[123,20,3],[40,145,8],[34,6,132]])

fig, ax = plt.subplots()
im = ax.imshow(M, cmap='gray_r', alpha=0.8)
fig.colorbar(im, ax=ax)

for i in range(3):
    for j in range(3):
        ax.text(j, i, str(M[i, j]), ha='center', va='center')
        
ax.set_xticks([0,1,2])
ax.set_yticks([0,1,2])
ax.set_xticklabels(['A','B','C'])
ax.set_yticklabels(['A','B','C'])
ax.xaxis.tick_top()

ax.set(xlabel='Predicted class')
ax.set(ylabel='Real class')
ax.xaxis.set_label_position('top')
```
