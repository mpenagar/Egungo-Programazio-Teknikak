## Datu Galduen Kudeaketa (Missing Values)


7. **Balio gabeak hautematea:** `isnull()` funtzioak balioren bat falta ote den (edo None, NaN ote den) conprobatzen du. Emaitza boolearra bada ere, eragikweta aritmetikoetan (`sum()`, `mean()` funtzioak adibidez) balio boolearrak erabil daitezke (`True==1`, `False==0`). Konprobatu ezazu zutabe bakoitzean falta diren balioen portzentaia (`df.columns` zutabe-izenak dira).

`df.isnull()` funtzioa DataFrame osoaren gain aplikatzean, DataFrame boolear bat bueltatzen du. `df.any()` funtzioak `True` elementuren bat ote dagoen konprobatzen du. Saiatu bilatzen bidaiari guztietatik daturen bat falta zaien portzentaia. Kontuz, `df.any()` funtzioak defektuz zutabeka kalkulatzen du, baina guri ilaraka kalkulatzea interesatzen zaigu). 

8. **Ordezkapen numerikoa (Mediana):** Ezaugarri kuantitatiboetan, balio galduak ezaugarriaren medianarekin (`median()`) ordezkatu daitezke `fillna()` erabiliz. Ordezkatu *Age* ezaugarriaren balio galduak bere medianarekin. Aurreko eta ondorengo ezaugarri horren balio galdu kopurua konprobatu. Konprobatu ere zein den orain daturen bat falta zaien bidaiarien portzentaia.

**9 - Ordezkapen kategorikoa (Moda):** Ezaugarri kategorikoetan, balio galduak ezaugarriaren modarekin (`mode()`) ordezkatu daitezke `fillna()` erabiliz. Ordezkatu *Embarked* ezaugarriaren balio galduak bere modarekin. Aurreko eta ondorengo balio galdu kopurua konprobatu. Konprobatu ere zein den orain daturen bat falta zaien bidaiarien portzentaia. Kontuz: datu multzo baten modan elementu bat baina gehiago egon litezke:

```python
pd.Series([1,3,1,2,3,4]).mode()
```
```
0    1
1    3
dtype: int64
```

Aurreko adibide honetan ikusten denez, `mode` funtzioaren emaitza bi balio dituen serie bat da (1 eta 3 balioak, 2 aldiz agertzen direnak). Hau dela eta, `mode()[0]` erabiliko dugu (modetatik lehenengo elementua)

10. **Ordezkapen arbitrarioa:** Kasu batzuetan, adibidez *Cabin* ezaugarrian, ez du zentzurik mediana/moda erabiltzeak. Halakoetan balio galduak guk aukeratutako balio batekin ordezkatu ditzakegu. Ordezkatu *Cabin* ezaugarriaren balio galduak `'UNK'`(*unknown*) balioarekin. Aurreko eta ondorengo balio galdu kopurua konprobatu. Konprobatu ere zein den orain daturen bat falta zaien bidaiarien portzentaia.  


## Ezaugarrien Ingeniaritza eta Eraldaketa (Feature Engineering)

**11 - Eragiketa bidezko zutabe berriak:** Sortu ezaugarri berri bat lehendik dauden bi zutaberen arteko kalkulu baten bidez (adib. `FamilyMembers = 1 + SibSp * Parch`).

**12 - Zutabe bitar berriak:** Sortu ezaugarri bitar berri bat (adib. `Adult = Age >= 18`) eta kalkulatu balio faltsuen (adin behekoen) kopurua. Ezezko logikoa `~` eragilearen bidez adierazten da.

**13 - One-Hot Encoding (`get_dummies`):** Bihurtu ezaugarri kategoriko bat (*Embarked* adibidez) azpi-zutabe binarizatuetan `pd.get_dummies(df, columns=[..])` erabiliz, `drop_first=True` gehituz *dummy variable trap* deritzona saihesteko eta `dtype=int` zenbakizko balioak lortzeko.

**14 - Diskretizazioa (`cut`):** `pd.cut` funtzioa erabiliz, sortu `'AgeGroup'` ezaugarri berri bat, `'Age'` ezaugarriaren balio diskretizatuekin: `'Child'<18, 'Young'<35, 'Middle-aged'<65, 'Senior'>=65` . `cut` funtzioaren erabilpen adibide bat:

```python
temp = pd.Series([18,14,23,35,-7,3,24,13,2])
bins = [-np.inf, 13, 22, np.inf]
labels = ['cold', 'warm', 'hot']
pd.cut(temp, bins=bins, labels=labels)
```
```
0    warm
1    warm
2     hot
3     hot
4    cold
5    cold
6     hot
7    cold
8    cold
dtype: category
Categories (3, str): ['cold' < 'warm' < 'hot']
```

15. **Ordinal Encoding (`map`):** Bihurtu ordenatutako ezaugarri kategoriko bat (gure kasuan *AgeGroup* adibidez) balio numerikoetan (0, 1, 2...) `.map()` funtzioa erabiliz. Funtzioaren erabilpen adibide bat:

```python
temp = pd.Series(['warm','warm','hot','hot','cold','cold','hot','cold','cold'])
mapping = {'cold':0, 'warm':1, 'hot':2}
temp.map(mapping)
```
```
0    1
1    1
2    2
3    2
4    0
5    0
6    2
7    0
8    0
dtype: int64
```
