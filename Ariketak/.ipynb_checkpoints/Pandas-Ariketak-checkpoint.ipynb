Ariketa multzo honek Pandas liburutegiaren oinarrizko lan-fluxua jorratzen du, Machine Learning (ML) proiektuetan ezinbestekoak diren datuen esplorazio, garbiketa, ezaugarri-ingeniaritza eta aurreprozesamendu prozesuetan ardatza jarriz.

**1. Datuak Kargatzea eta Lehen Esplorazioa (EDA)**

1. **CSV karga eta dimentsioak:** Kargatu CSV fitxategi bat DataFrame batean eta bistaratu bere neurria (`shape`), zutabeen datu-motak (`dtypes`) eta lehenengo 5 errenkadak (`head()`).
2. **Laburpen estatistikoa:** Sortu ezaugarri numerikoen laburpen estatistikoa (`.describe()`) eta aztertu batez bestekoak, kuartilak eta balio minimo/maximoak.
3. **Bikoiztuak kudeatzea:** Aurkitu datu-multzoan bikoiztutako errenkada kopurua (`duplicated().sum()`) eta ezabatu itzazu DataFrame-tik (`drop_duplicates()`).

**2. Hautaketa, Iragazketa eta Slicing**

4. **Ezaugarriak ($X$) eta Etiketa ($y$) bereiztea:** Erabili `iloc` edo `loc` azken zutabea helburu-bektore gisa ($y$) eta gainerako zutabe guztiak ezaugarri-matrize gisa ($X$) hautatzeko.
5. **Iragazketa baldintzatu anitza:** Iragazi DataFrame-a bi baldintza logiko aldi berean betetzen dituzten errenkadak lortzeko (adib. `Adina > 30` eta `Soldata < 50000`).
6. **Kategorien araberako hautaketa (`isin`):** Hautatu zutabe kategorial batean zerrenda jakin bateko balioak dituzten errenkadak soilik (adib. `Hiria.isin(['Bilbo', 'Gasteiz'])`).

**3. Datu Galduen Kudeaketa (Missing Values)**

7. **Balio gabeak hautematea:** Kalkulatu zutabe bakoitzeko balio nuluen kopurua eta ehunekoa (`isnull().mean() * 100`).
8. **Imputazio numerikoa (Mediana):** Bete zutabe numeriko baten balio galduak bere medianarekin (`fillna()`), ML ereduek balio nulurik onartzen ez dutelako.
9. **Imputazio kategoriala (Moda):** Bete zutabe kategorial baten balio galduak balio arruntenarekin (`mode()[0]`).
10. **Atalase bidezko ezabatzea:** Ezabatu balio nuluen %40 baino gehiago dituzten zutabe guztiak (`dropna(thresh=...)`).

**4. Ezaugarrien Ingeniaritza eta Eraldaketa (Feature Engineering)**

11. **Eragiketa bidezko zutabe berriak:** Sortu ezaugarri berri bat lehendik dauden bi zutaberen arteko kalkulu baten bidez (adib. `Guztira = Kantitatea * Prezioa`).
12. **Binarizazioa:** Bihurtu ezaugarri numeriko bat kategorial binarizatuan (0 edo 1) `np.where()` edo baldintza logiko baten bidez (adib. `Adina >= 18`).
13. **One-Hot Encoding (`get_dummies`):** Bihurtu zutabe kategorial nominal bat azpi-zutabe binarizatuetan `pd.get_dummies()` erabiliz, *dummy variable trap* saihesteko `drop_first=True` gehituz.
14. **Ordinal Encoding (`map`):** Bihurtu ordenatutako kategoria testualak ('Baxua', 'Ertaina', 'Altua') balio numerikoetan (0, 1, 2) `.map()` hiztegi bat erabiliz.
15. **Balio atipikoak moztea (Outlier Clipping):** Mugatu zutabe baten goiko eta beheko balio muturrekoak percentil zehatz batzuetan `.clip()` funtzioa erabiliz.

**5. Taldekatzea eta Agregazioa (Grouping & Aggregation)**

16. **Batez bestekoak taldeka:** Taldekatu datuak ezaugarri kategorial baten arabera (`groupby`) eta kalkulatu zutabe numerikoen batez bestekoa.
17. **Agregazio anitzak (`agg`):** Aplikatu agregazio-funtzio ezberdinak zutabe ezberdinei aldi berean (adib. `Adina`-ri mediana eta `Soldata`-ri batez bestekoa zein kopurua).
18. **Pibote-taulak (Pivot Tables):** Sortu pibote-taula bat (`pivot_table`) bi ezaugarri kategorialen arteko gurutzaketa eta hirugarren balio numeriko baten metrika aztertzeko.

**6. Datasets Batzea eta Exportatzea**

19. **DataFrames Batzea (`merge`):** Batu bi DataFrame ezberdin gako-zutabe komun baten bidez (`pd.merge`), barne-lotura (*inner join*) eginez.
20. **Garbiketaren ondorengo eksportazioa:** Gorde eraldatutako eta garbitutako azken DataFrame-a CSV fitxategi berri batean (`to_csv(index=False)`), ML eredua entrenatzeko erabilgarri utziz.
