# NumPy

Ariketa multzo honek NumPy liburutegiaren oinarrizko eragiketak lantzen ditu, Ikaskuntza Automatikoan (ML) ohikoak diren datu-egitura, matrize-eragiketa eta aurreprozesamendu tekniketan arreta jarriz.

## Bektore/matrize Sorrera eta Forma Aldaketa

**1 - Bektoreen hasieratzea:** Sortu 100 zeroz osatutako 1D array bat.

**2 - Pisu-matrizeen hasieratzea:** Sortu 1ez betetako $5 \times 3$ neurriko matrize bat.

**3 - Sekuentzia lineala:** Sortu 10etik 50era bitarteko zenbaki bikoitien array bat (`np.arange` erabiliz).

**4 - Espazio jarraitua:** Sortu 0 eta 1 artean berdinki banatutako 20 zenbakiz osatutako array bat (`np.linspace` erabiliz).

**5 - Dimentsioak eta Iraulia:** Sortu 12 elementuko 1D array bat, bihurtu $(4, 3)$ tamainako ezaugarri-matrize eta kalkulatu bere iraulia ($(3, 4)$ tamainako matrizea).

**6 - Flattening / Unrolling:** Bihurtu $3 \times 3$ neurriko matrize bat 1D bektore linealean (`flatten` edo `ravel` erabiliz), irudi baten pixelak ezaugarri-bektore bihurtzeko eragiketa klasikoa.

## Indizeak, Mozketak (Slicing) eta Maskara Boolearrak

**7 - Data-Split simulazioa ($X$ eta $y$):** $5 \times 4$ neurriko matrize bat izanda, banatu lehenengo 3 zutabeak ezaugarrien matrizean ($X$) eta azken zutabea helburu-bektorean ($y$).

**8 - Azpi-multzoak erauztea:** Lortu (erauzi) $4 \times 4$ matrize baten erdialdeko $2 \times 2$ azpimatrizea.

**9 - Binarizazioa (Atalase-funtzioa):** Sortu 20 ausazko zenbakiko array bat (0 eta 1 artean). Maskara boolearra erabiliz, ordezkarazi 0.5 baino handiagoak diren balioak 1ekin eta txikiagoak 0rekin.

**10 - ReLU Aktibazio-funtzioa:** Sortu balio positibo zein negatiboak dituen 1D bektore bat, eta ordezkarazi balio negatibo guztiak 0rekin (`np.maximum` edo maskara boolearra erabiliz).

## Operazio Matematikoak eta Estatistika

**11 - Ardatzen arteko agregazioa:** $(5, 3)$ neurriko matrize baten batez bestekoa eta desbiderapen estandarra kalkulatu zutabe bakoitzeko (`axis=0`) eta errenkada bakoitzeko (`axis=1`).

**12 - Min-Max Eskalatzea (Normalization):** Normalizatu 1D array bat balio guztiak $[0, 1]$ tartean egon daitezen formula hau erabiliz:

$$\frac{x - x_{min}}{x_{max} - x_{min}}$$

**13 - Min-Max Eskalatzea Matrizeetan:** Aurreko formula erabiliz, normalizatu matrize batetako zutabe bakoitza (datu ezaugarri bakoitza).

**14 - Broadcasting bidezko estandarizazioa:** $(5, 3)$ matrize bati kendu zutabe bakoitzaren batez besteko bektorea (Broadcasting gaitasuna erabiliz ziurtatu eragiketa bektoriala dela).


**15 - Sailkapen anitzeko iragarpena (Argmax):** $(5, 3)$ neurriko matrize batek 5 laginen 3 klase posibleen probabilitateak adierazten ditu. Aurkitu lagin bakoitzari dagokion klase iragarriaren indizea (`np.argmax` erabiliz `axis=1` ardatzean).

## Aljebra Lineala eta Biderketak

**16 - Biderkadura eskalarra (Dot Product):** Kalkulatu ezaugarri-bektore baten ($x$, 4 luzerakoa) eta pisu-bektore baten ($w$, 4 luzerakoa) arteko biderkadura eskalarra ($w \cdot x$).

**17 - Matrize-biderketa (Batch iragarpena):** Biderkatu $X$ datu-multzoa ($10 \times 4$ neurrikoa) eta $W$ pisu-matrizea ($4 \times 2$ neurrikoa) iragarpen-matrizea lortzeko (`@` eragilea edo `np.dot` erabiliz).

**18 - Distantzia Euklidearra (k-NN oinarria):** Kalkulatu bi 1D bektoreren arteko distantzia euklidearra:

$$d = \sqrt{\sum_{i=1}^{n} (a_i - b_i)^2}$$

## Ausazko Zenbakiak eta Laginketa

**19 - Pisuen Ausazko Hasieratzea:** Erabili `np.random.seed(42)` birsortzea ziurtatzeko, eta sortu banaketa normal estandarra ($\mu=0, \sigma=1$) jarraitzen duen $(4, 3)$ neurriko pisu-matrize bat.

**20 - Data Shuffling (Indizeak nahastea):** Sortu 0tik 99ra arteko indize-bektore bat eta nahastu ausaz (`np.random.shuffle`), datu-multzo bat entrenamendu/test multzoetan banatu aurretik nahasteko.

**21 - Sigmoide Funtzio bektoriala:** Inplementatu NumPy erabiliz Sigmoide aktibazio-funtzioa eta aplikatu ausaz sortutako 1D bektore bati:

$$\sigma(z) = \frac{1}{1 + e^{-z}}$$
