### Ikaskuntza Automatikoa:
Ordenagailuek **datuetatik ikastea** ahalbidetzen duen zientzia



"**T ataza** bat eta hari lotutako **P errendimendu-neurri** bat izanik, ordenagailu-programa batek **E esperientzia** batetik <u>ikasten du</u>, baldin eta T atazan duen errendimenduak (P-ren bidez neurtuta) E esperientziaren poderioz hobera egiten badu". Tom Mitchell (1997)

### Ikaskuntza Automatikoa: T Ataza

* $\mathbf{X}$ &nbsp; **aldagai iragarlea**: sarrerako datuak, ezaugarriak edo behaketak
* $\mathbf{Y}$ &nbsp; **helburu-aldagaia**: erantzuna edo iragarri nahi dugun informazioa

Ataza ebazteko, programak

$$f: \mathbf{X} \rightarrow \mathbf{Y}$$

egitura duen funtzio matematiko bat eraikitzen du.



### Ikaskuntza Automatikoaren Taxonomia
$\mathbf{Y}$ helburu-aldagaiaren izaeraren arabera, ataza mota ezberdinak defini daitezke.

* **Erregresioa**: Izaera kuantitatiboa eta ordenatua (**ZENBAT?**)

* **Sailkapena**: Izaera kategorikoa (**ZER?**)

* **Multzokatzea**: Izaera kategorikoa, aurrez ezezaguna (**ZER?**)

* **Kokapena/Detekzioa**: Laukizuzen baten koordenatu espazialak (**NON?**)

* **Jarraipena**": Laukizuzen baten koordenatu espazialen eboluzioa (**NONDIK?**)

### Ikaskuntza Automatikoa: E Esperientzia

Ikaskuntzarako erabiltzen den $E$ esperientzia datu multzo bat izan ohi da, ebatzi nahi den atazaren adibideez osotua:

$$E = \{(x_1, y_1), (x_2, y_2), \dots, (x_N, y_N)\} = \{(x_i, y_i)\}_{i=1}^N$$

* $x_i$: behaketa edo instantzia (aldagai iragarlea)
* $y_i$: etiketa, balioa edo erantzuna (helburu-aldagaia)



### Ikaskuntza Automatikoaren Taxonomia II
$E$ esperientzia datu multzoaren izaeraren arabera, ikaskuntza mota ezberdinak defini daitezke.

* **Gainbegiratua**: $E$ multzoko elementuetan aldagai iragarlea eta helburu-aldagaia agertzen dira beti. Elementu guztiak etiketatuak daude, $E = \{(x_i, y_i)\}_{i=1}^N$

* **Erdi-gainbegiratua**: $E$ multzoko elementu batzuetan ez da helburu-aldagairik agertzen. Elementu batzuk daude soilik etiketatuak, $E = \{(x_i, y_i)\}_{i=1}^K \cup \{(x_i)\}_{i=K+1}^N$

* **Ez-gainbegiratua**: $E$ multzoko elementuek ez dute helburu-aldagairik. Ez dago elementu etiketaturik, $E = \{(x_i)\}_{i=1}^N$  

## 1.3 - Ikaskuntza Automatikoa - Funtsezko osagaiak 

### Datu-multzoa

**Datu-multzoa** (*Dataset*): Eredua elikatzeko erabiltzen den informazio bilduma, askotan taula gisa adierazia

<img src="img/dataset.png" style="display: block; margin: 0 auto; width: 40%; height: auto;" alt="Dataset table"/>

* $(\mathbf{x_i}, y_i)\,$: lagina (taulako ilara bat)
* $\mathbf{x_i}\,$: ezaugarri bektorea, aldagai iragarlearen balioa
* $y_i\,$: helburu-aldagaiaren balioa
   * Sailkapen-ataza: Etiketa (*Label*)
   * Erregresio-ataza: Erantzuna (*Target*)

Datu multzoak ebazten ari garen ataza errepresentatzen du.
* **Nondik** datoz? **Nola** sortu (lagindu) da datu-multzoa?
* Datu kantitatea **nahikoa** al da?
* Zer nolako **zarata** dago datuetan?
* **Laginketa-alborapenik** (*sesgo de muestreo*) egon liteke?
* Ezaugarri guztiak **egokiak** edo **beharrezkoak** al dira?

### Eredua

Eredu bat, $\mathbf{x}$ ezaugarri bektoreen eta $y$ helburu-aldagaiaren arteko erlazioa bere gain hartzen duen funtzio matematiko bat da:

$$f: \mathbf{X} \rightarrow \mathbf{Y}$$

Praktikan, eredu-funtzioa zehazteko parametro-bektore bat erabili ohi da:

$$\hat{y} = f(\mathbf{x}; \boldsymbol{\theta})$$

* $\mathbf{x}\,$: Sarrerako ezaugarri-bektorea.
* $\hat{y}\,$: Ereduak emandako estimazioa edo iragarpena (benetako $y$ balioaren hurbilpena).
* $\boldsymbol{\theta}\,$: Doitzen diren ereduaren barne-parametroak.

### Ikaskuntza prozesuaren atal nagusiak

**Entrenamendua** (*Training / Fit*):
* Ereduak datuetatik ikasten duen fasea.
* Makinari $\mathbf{x}$ ezaugarriak eta $y$ emaitza zuzenak ematen zaizkio, bien arteko erlazio matematikoak (patroiak) aurkitu/erauzi ditzan.

**Inferentzia edo Iragarpena** (*Inference / Predict*):
* Eredua martxan jartzeko fasea.
* Ikasitako arauak aplikatzen dira $\mathbf{x}$ sarrerako datu berri batean, haren $y$ helburu-aldagai ezezaguna iragartzeko.

### Ikaskuntza-helburua - Optimizazioa

Entrenamenduaren xedea testuinguru honetan $\boldsymbol{\theta}^*$ parametro optimoak aurkitzea da, errorea edo galerak gutxitzeko:

$$\boldsymbol{\theta}^* = \arg\min_{\boldsymbol{\theta}} \sum_{i=1}^{N} \mathcal{L}(y_i, \hat{y}_i) = \arg\min_{\boldsymbol{\theta}} \sum_{i=1}^{N} \mathcal{L}(y_i, f(\mathbf{x}_i; \boldsymbol{\theta}))$$

* $\mathcal{L}\,$: benetako $y_i$ balioaren eta $\hat{y}_i$ iragarpenaren arteko desbideraketa neurtzen duen **Galera-funtzioa**

### Datuen banaketa (Nola/Non neurtu errendimendua?)

* **Entrenamendu-multzoa** (*Training set*): Datu-multzoaren zatirik handiena, eredua doitzeko soilik erabiltzen dena.

* **Test-multzoa** (*Test set*): Ereduaren benetako zehaztasuna ebaluatzeko berariaz gordetzen den datu-zatia (ereduak ikaskuntza-fasean **<u>sekula</u>** ikusi ez duena).

<u>Bekatu nagusia</u>: Eredu bat entrenamendurako erabilitako datu berekin ebaluatzea. Ezingo dugu jakin ezaugarrien eta erantzunaren arteko erlazioa benetan barneratu duen (orokortu duen) edo besterik gabe buruz ikasi dituen.

## 1.4 - Ikaskuntza Automatikoaren erronkak

### Orokortzea (*Generalization*)

Ereduak inoiz ikusi ez dituen datu berrien aurrean asmatzeko duen gaitasuna da orokortzea. Hori gabe, ereduak ez du ezertarako balio.

* **Seinalea**: Datuen atzean dagoen egitura, arau edo joera egiazkoa (ikasi nahi duguna).
* **Zarata**: Datuetako ausazko gorabeherak, salbuespenak edo neurtze-erroreak (alboratu nahi duguna).

Eredu on bat **<u>seinalearekin</u>** geratzen da, zarata baztertuz.

### Ikaskuntzaren bi muturrak

**Azpidoikuntza** (*Underfitting*):
* Eredua sinpleegia da eta ez da gai datuetan ezkutuan dagoen egitura antzemateko.
* <u>Ondorioa</u>: Akats asko egiten ditu bai entrenamendu-datuekin, bai test-datuekin.

**Gaindoikuntza** (*Overfitting*):
* Eredua konplexuegia da eta egitura ulertu beharrean, entrenamendu-datuak buruz ikasten ditu, zarata bera ere egitura balitz bezala barneratuz.
* <u>Ondorioa</u>: Akatsik ez entrenamenduan (nota bikaina), baina izugarrizko porrota datu berriekin (ezin du orokortu).

<img src="img/underfitting-overfitting.png" style="display: block; margin: 0 auto; width: 100%; height: auto;" alt="underfitting vs overfitting"/>


### Alborapen-Bariantza Oreka (*Bias-Variance Tradeoff*)

Eredu baten errorea bi iturri nagusitatik dator eta ezin dira biak aldi berean zerora eraman:

* **Alborapena** (*Bias*): Ereduaren sinpletasunak sortzen duen errorea. Alborapen handiko eredu batek hipotesi zurrunegiak ditu (adibidez, dena lerro zuzen bat dela pentsatzea). Azpidoikuntzari lotuta dago.

* **Bariantza** (*Variance*): Ereduak entrenamendu-datuen aldaketa txikiekiko duen sentikortasuna. Bariantza handiko eredu bat asko aldatzen da datu batzuk kendu edo gehituz gero. Gaindoikuntzari lotuta dago.

<u>Dilema</u>: Ereduaren konplexutasuna handitzean, alborapena murrizten da, baina bariantza handitu. Gure helburua bien arteko oreka aurkitzea da.

## 1.6 - Ikaskuntza Automatikoko aplikazioen adibideak

* **Spam-posta detekzioa**: Posta elektroniko baten edukia aztertu eta spama ote den iragartzea.

* **Inmobiliaria-merkatuko prezioen estimazioa**: Etxe baten ezaugarriak erabiliz (azalera, kokapena, gela kopurua), haren salmenta-prezioa iragartzea.

* **Iruzurren detekzioa banku-transakzioetan**: Txartel baten mugimendua legezkoa ote den iragartzea sarrera-ezaugarrietan oinarrituta (kopurua, kokapena, ordua, ...).

* **Sentimenduen analisia albiste edo sare sozialetan**: Enpresa bati buruz idatzitako iruzkin edo albiste bat positiboa, negatiboa ala neutroa den iragartzea.

* **Minbizi-detekzioa irudi medikoetan**: X izien edo erresonantzien bidez, ehun bat osasuntsua den edo tumore mota zehatz bat dakarren iragartzea.

* **Trafiko-seinaleen kudeaketa ibilgailu autonomoetan**: Kamerak hartutako irudi bakoitzean agertzen diren gida-seinaleak iragartzea.

* **Eskariaren aurreikuspena (Demand Forecasting)**: Saltoki batean datorren astean produktu jakin baten zenbat unitate salduko diren iragartzea, eguraldiaren eta joeren arabera.

* **Edukiaren gomendioa**: Netflix edo Spotify bezalako plataformetan, erabiltzaile baten iraganeko portaera ikusita, gustatuko lizaion hurrengo filma edo abestia proposatzea.

* **Anomalia-detekzioa sentsore telemetrikoetan**: Fabrika batetan ohiko portaeratik kanpo gelditzen den makina bat identifikatzea, mantentze-lanak egiteko.

* **Itzulpen automatikoa**: Testu bat hizkuntza batetik bestera itzultzea.

* **Bezeroen segmentazioa marketinean**: Erosketa-ohiturak alderatuz, bezeroak talde ezberdinetan antolatzea, talde bakoitzari kanpaina pertsonalizatuak bidaltzeko.
